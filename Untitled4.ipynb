```python
import numpy as np
from scipy.ndimage import gaussian_filter
from PIL import Image
import matplotlib.pyplot as plt
from itertools import product

class UnionFind:
    def __init__(self, size):
        self.parent = list(range(size))
        self.size = [1] * size
        self.max_edge = [0] * size

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]

    def union(self, x, y, weight):
        root_x = self.find(x)
        root_y = self.find(y)

        if root_x == root_y:
            return

        if self.size[root_x] < self.size[root_y]:
            root_x, root_y = root_y, root_x

        self.parent[root_y] = root_x
        self.size[root_x] += self.size[root_y]
        self.max_edge[root_x] = max(self.max_edge[root_x], self.max_edge[root_y], weight)

def build_graph(img_gray):
    height, width = img_gray.shape
    edges = []
    
    # Generate 8-connected neighborhood
    offsets = list(product([-1, 0, 1], repeat=2))
    offsets.remove((0, 0))
    
    for y in range(height):
        for x in range(width):
            current_idx = y * width + x
            current_val = img_gray[y, x]
            
            for dy, dx in offsets:
                ny, nx = y + dy, x + dx
                if 0 <= ny < height and 0 <= nx < width:
                    neighbor_idx = ny * width + nx
                    if neighbor_idx > current_idx:  # Avoid duplicate edges
                        # Grayscale difference (absolute intensity difference)
                        diff = abs(current_val - img_gray[ny, nx])
                        edges.append((diff, current_idx, neighbor_idx))
    
    return edges

def segment_grayscale(img_gray, sigma=0.5, k=300, min_size=50):
    # Convert to float32 and add channel dimension if needed
    if img_gray.ndim == 3:
        img_gray = img_gray[:, :, 0]
    
    # Preprocess image (single-channel Gaussian smoothing)
    smoothed = gaussian_filter(img_gray.astype(np.float32), sigma=sigma)
    
    height, width = smoothed.shape
    total_pixels = height * width
    
    # Build and sort edges
    edges = build_graph(smoothed)
    edges.sort(key=lambda x: x[0])
    
    # Initialize union-find
    uf = UnionFind(total_pixels)
    
    # Process edges
    for weight, u, v in edges:
        root_u = uf.find(u)
        root_v = uf.find(v)
        
        if root_u != root_v:
            threshold_u = uf.max_edge[root_u] + k / uf.size[root_u]
            threshold_v = uf.max_edge[root_v] + k / uf.size[root_v]
            
            if weight <= min(threshold_u, threshold_v):
                uf.union(u, v, weight)
    
    # Generate label matrix
    labels = np.zeros((height, width), dtype=np.int32)
    for y in range(height):
        for x in range(width):
            labels[y, x] = uf.find(y * width + x)
    
    return labels

# Usage example
image = np.array(Image.open('IAW3.JPG').convert('L'))  # Convert to grayscale
segmented = segment_grayscale(image, sigma=2.0, k=3000)

# Visualize results
plt.imshow(segmented, cmap='nipy_spectral')
plt.show()
```

```python
import numpy as np
from scipy.ndimage import gaussian_filter
import matplotlib.pyplot as plt
from itertools import product
from pyhdf.SD import SD, SDC  # Requires pyhdf installation
import cv2 as cv2

class UnionFind:
    def __init__(self, size):
        self.parent = list(range(size))
        self.size = [1] * size
        self.max_edge = [0] * size

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]

    def union(self, x, y, weight):
        root_x = self.find(x)
        root_y = self.find(y)

        if root_x == root_y:
            return

        if self.size[root_x] < self.size[root_y]:
            root_x, root_y = root_y, root_x

        self.parent[root_y] = root_x
        self.size[root_x] += self.size[root_y]
        self.max_edge[root_x] = max(self.max_edge[root_x], 
                                  self.max_edge[root_y], weight)

def build_graph(img_gray):
    height, width = img_gray.shape
    edges = []
    
    # 8-connected neighborhood offsets
    offsets = [(dy, dx) for dy in (-1,0,1) for dx in (-1,0,1) 
              if not (dy == 0 and dx == 0)]
    
    for y in range(height):
        for x in range(width):
            current_idx = y * width + x
            current_val = img_gray[y, x]
            
            for dy, dx in offsets:
                ny, nx = y + dy, x + dx
                if 0 <= ny < height and 0 <= nx < width:
                    neighbor_idx = ny * width + nx
                    if neighbor_idx > current_idx:
                        diff = abs(current_val - img_gray[ny, nx])
                        edges.append((diff, current_idx, neighbor_idx))
    return edges

def segment_hdf(filename, dataset_index=0, sigma=0.8, k=100, min_size=50):
    # Read HDF dataset
    hdf = SD(filename, SDC.READ)
    try:
        dataset_name = list(hdf.datasets().keys())[dataset_index]
        dataset = hdf.select(dataset_name)
        data = dataset[:]
        
        # Handle 3D data (select first element from first dimension)
        if data.ndim == 3:
            data_fed = data[0, 200:850,0 :1023]
            data_min_val = data_fed.min()
            data_max_val = data_fed.max()
            data_norm = ((data_fed - data_min_val)/(data_max_val - data_min_val))*255.0
            data3 = data_norm.astype(np.float32)
            img_gray = data3
        else:
            img_gray = data_norm.astype(np.float32)
            
    finally:
        hdf.end()

        

    # Preprocessing
    smoothed = gaussian_filter(img_gray, sigma=sigma)
    height, width = smoothed.shape
    total_pixels = height * width
    
    # Build and sort edges
    edges = build_graph(smoothed)
    edges.sort(key=lambda x: x[0])
    
    # Initialize union-find
    uf = UnionFind(total_pixels)
    
    # Process edges
    for weight, u, v in edges:
        root_u = uf.find(u)
        root_v = uf.find(v)
        
        if root_u != root_v:
            threshold_u = uf.max_edge[root_u] + k / uf.size[root_u]
            threshold_v = uf.max_edge[root_v] + k / uf.size[root_v]
            
            if weight <= min(threshold_u, threshold_v):
                uf.union(u, v, weight)
    
    # Generate labels
    labels = np.zeros((height, width), dtype=np.int32)
    for y in range(height):
        for x in range(width):
            labels[y, x] = uf.find(y * width + x)
            
    labels_normalized = cv2.normalize(labels, None, 0, 255, cv2.NORM_MINMAX, dtype=cv2.CV_8U)
    
    return labels_normalized

def data_analysis(img):
    # Convert to proper OpenCV format if needed
    if img.dtype != np.uint8:
        img = cv2.normalize(img, None, 0, 255, cv2.NORM_MINMAX, dtype=cv2.CV_8U)
    
    # Create working copy for visualization
    display_img = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)
    
    # Find corners in the segmented image
    corners = cv2.goodFeaturesToTrack(img, 500, 0.001, 20)
    if corners is None:
        return 0, img.shape[1], 0, img.shape[0]  # Return full image bounds if no corners
    
    corners = np.intp(corners).reshape(-1, 2)

    # Filter corners with neighbor check
    filtered_corners = []
    max_distance = 50
    for i, corner in enumerate(corners):
        for j, other_corner in enumerate(corners):
            if i != j and np.linalg.norm(corner-other_corner) <= max_distance:
                filtered_corners.append(corner)
                break
    
    # Draw filtered corners on display image
    filtered_corners = np.array(filtered_corners)
    for corner in filtered_corners:
        x, y = corner.ravel()
        cv2.circle(display_img, (x, y), 5, (0, 255, 0), 2)

    # Calculate bounding box with safety margin
    if len(filtered_corners) > 0:
        min_x = filtered_corners[:, 0].min() - 30
        max_x = filtered_corners[:, 0].max() + 30
        min_y = filtered_corners[:, 1].min() - 30
        max_y = filtered_corners[:, 1].max() + 30
    else:  # Fallback if no corners remain after filtering
        min_x, max_x, min_y, max_y = 0, img.shape[1], 0, img.shape[0]

    # Clamp values to image dimensions
    h, w = img.shape[:2]
    min_x = max(0, min_x)
    max_x = min(w, max_x)
    min_y = max(0, min_y)
    max_y = min(h, max_y)

    # Display results
    cv2.imshow('Segmentation with Corners', display_img)
    cv2.waitKey(1)  # Small delay to allow window update
    
    return min_x, max_x, min_y, max_y

# Integrated workflow
def full_analysis_pipeline(filename, dataset_index=0):
    # Step 1: Perform segmentation
    segmented = segment_hdf(filename, dataset_index)
    
    # Step 2: Analyze segmented image
    min_x, max_x, min_y, max_y = data_analysis(segmented)
    
    # Step 3: Load original data for cropping
    hdf = SD(filename, SDC.READ)
    try:
        dataset_name = list(hdf.datasets().keys())[dataset_index]
        dataset = hdf.select(dataset_name)
        original_data = dataset[:]
    finally:
        hdf.end()
    
    # Crop original data using analysis results
    if original_data.ndim == 3:
        cropped = original_data[:, min_y:max_y, min_x:max_x]
    else:
        cropped = original_data[min_y:max_y, min_x:max_x]
    
    return cropped, (min_x, max_x, min_y, max_y)

# Example usage
filename = 'data/IAW112131.hdf'
result, bounds = full_analysis_pipeline(filename)

# Visualization
plt.figure(figsize=(12, 6))
plt.subplot(121)
plt.imshow(result[0] if result.ndim == 3 else result, cmap='gray')
plt.title('Cropped Region')
plt.subplot(122)
plt.imshow(cv2.normalize(segment_hdf(filename), None, 0, 255, cv2.NORM_MINMAX))
plt.gca().add_patch(plt.Rectangle((bounds[0], bounds[2]), 
                   bounds[1]-bounds[0], bounds[3]-bounds[2],
                   fill=False, edgecolor='red', linewidth=2))
plt.title('Original with Bounding Box')
plt.show()



```
